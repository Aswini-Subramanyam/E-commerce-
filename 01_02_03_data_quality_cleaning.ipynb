## DATA CLEANING

```python
#Cleaning customers dataframe
def clean_customers(df: pd.DataFrame) -> pd.DataFrame:
    """
    Cleans the customers DataFrame:
    - Standardizes ZIP prefix to 5 digits (string)
    - Normalizes city (lowercase, stripped)
    - Normalizes state (uppercase, stripped)
    - Validates state length = 2
    - Removes helper column after check
    - Drops duplicate customer_id rows
    """
    df_clean = df.copy()

    # ZIP prefix as 5-digit string
    df_clean["customer_zip_code_prefix"] = (
        df_clean["customer_zip_code_prefix"].astype(str).str.zfill(5)
    )

    # Normalize city and state
    df_clean["customer_city"] = df_clean["customer_city"].str.lower().str.strip()
    df_clean["customer_state"] = df_clean["customer_state"].str.upper().str.strip()

    # Validate state length
    df_clean["state_length"] = df_clean["customer_state"].str.len()
    print(f"{df_clean["state_length"].value_counts()}\n")
    df_clean.drop(columns=["state_length"], inplace=True)

    # Drop duplicates by customer_id
    df_clean = df_clean.drop_duplicates(subset=["customer_id"])

    return df_clean

df_customers_clean = clean_customers(df_customers)

print(f"{df_customers_clean.dtypes}\n")
print(f"Unique customers: {df_customers_clean["customer_id"].nunique()}\n")
print(f"Shape: {df_customers_clean.shape}\n")
print(df_customers_clean.head())
```

```python
#Analysing seller dataframe
from unidecode import unidecode
import re

def clean_sellers(df: pd.DataFrame) -> pd.DataFrame:
    """
    Cleans the sellers DataFrame:
    - Normalizes seller_city (lowercase, accents removed, stripped, cleaned of noise)
    - Replaces purely numeric city values with NaN
    - Normalizes seller_state (uppercase, stripped)
    - Normalizes seller_zip_code_prefix to 5-digit strings
    - Fixes specific known issue: 'lages  sc' → 'lages'
    - Drops duplicate seller_id rows
    """
    df_clean = df.copy()

    # Normalize seller_city
    df_clean["seller_city"] = (
        df_clean["seller_city"]
        .apply(lambda x: unidecode(str(x)).lower().strip())
        .str.replace(r"\s+", " ", regex=True)      # collapse spaces
        .str.replace(r"\(.*?\)", "", regex=True)   # remove parentheses content
        .str.replace(r"/.*", "", regex=True)       # remove anything after slash
        .str.replace(r"@.*", "", regex=True)       # remove email domains
        .str.replace(r"[^a-z\s]", "", regex=True)  # keep only letters + spaces
        .str.strip()
    )

    # Replace purely numeric city values with NaN
    df_clean.loc[df_clean["seller_city"].str.isnumeric(), "seller_city"] = None

    # Normalize seller_state
    df_clean["seller_state"] = (
        df_clean["seller_state"]
        .apply(lambda x: unidecode(str(x)).upper().strip())
    )

    # Normalize seller_zip_code_prefix to 5-digit strings
    df_clean["seller_zip_code_prefix"] = (
        df_clean["seller_zip_code_prefix"].astype(str).str.strip().str.zfill(5)
    )

    # Fix specific case
    df_clean["seller_city"] = df_clean["seller_city"].replace({"lages  sc": "lages"})

    # Drop duplicate sellers by seller_id
    df_clean = df_clean.drop_duplicates(subset=["seller_id"])

    return df_clean

df_sellers_clean = clean_sellers(df_sellers)

print(f"{df_sellers_clean.dtypes}\n")
print(f"Unique customers: {df_sellers_clean["seller_id"].nunique()}\n")
print(f"Shape: {df_sellers_clean.shape}\n")
print(df_sellers_clean.head())
```

```python
from unidecode import unidecode

# -------------------------------
# Step 1: Clean Geolocation Table
# -------------------------------
def clean_geolocation(df: pd.DataFrame) -> pd.DataFrame:
    df_clean = df.copy()

    # Normalize ZIP prefix to 5-digit strings
    df_clean["geolocation_zip_code_prefix"] = (
        df_clean["geolocation_zip_code_prefix"].astype(str).str.zfill(5)
    )

    # Normalize state
    df_clean["geolocation_state"] = df_clean["geolocation_state"].str.upper().str.strip()

    # Normalize city
    df_clean["geolocation_city"] = (
        df_clean["geolocation_city"]
        .apply(lambda x: unidecode(str(x)).lower().strip().replace(".", ""))
    )

    # Drop duplicates
    df_clean = df_clean.drop_duplicates()

    return df_clean


# -------------------------------
# Step 2: Aggregate by ZIP Prefix
# -------------------------------
def aggregate_geolocation(df: pd.DataFrame) -> pd.DataFrame:
    return (
        df.groupby("geolocation_zip_code_prefix", as_index=False)
        .agg(
            lat=("geolocation_lat", "median"),
            lng=("geolocation_lng", "median"),
            city=("geolocation_city", lambda x: x.mode().iloc[0] if not x.mode().empty else None),
            state=("geolocation_state", lambda x: x.mode().iloc[0] if not x.mode().empty else None),
        )
    )


# -------------------------------
# Step 3: Build Summary for Missing ZIPs
# -------------------------------
def build_summary(df_customers, df_geo, window=10):
    # Normalize ZIPs
    df_customers["customer_zip_code_prefix"] = df_customers["customer_zip_code_prefix"].astype(str).str.zfill(5)
    df_geo["geolocation_zip_code_prefix"] = df_geo["geolocation_zip_code_prefix"].astype(str).str.zfill(5)

    # Find missing ZIPs
    missing_mask = ~df_customers["customer_zip_code_prefix"].isin(df_geo["geolocation_zip_code_prefix"])
    missing_zips = df_customers.loc[missing_mask, "customer_zip_code_prefix"].unique()

    summary_rows = []
    for zip_code in missing_zips:
        # Get city/state from customers
        cust_info = df_customers[df_customers["customer_zip_code_prefix"] == zip_code]
        city = cust_info["customer_city"].mode().iloc[0]
        state = cust_info["customer_state"].mode().iloc[0]

        # Get lat/lon from neighbors
        zip_int = int(zip_code)
        neighbors = df_geo[df_geo["geolocation_zip_code_prefix"].astype(int).between(zip_int - window, zip_int + window)]
        if not neighbors.empty:
            lat = neighbors["lat"].median()
            lng = neighbors["lng"].median()
        else:
            lat, lng = None, None

        summary_rows.append({
            "geolocation_zip_code_prefix": zip_code,
            "lat": lat,
            "lng": lng,
            "city": city,
            "state": state
        })

    return pd.DataFrame(summary_rows)


# -------------------------------
# Step 4: Build Summary for Sellers Missing ZIPs
# -------------------------------
def build_summary_sellers(df_sellers, df_geo, missing_zips, window=10):
    summary_rows = []
    for zip_code in missing_zips:
        # Get city/state from sellers
        seller_info = df_sellers[df_sellers["seller_zip_code_prefix"] == zip_code]
        city = seller_info["seller_city"].mode().iloc[0]
        state = seller_info["seller_state"].mode().iloc[0]

        # Get lat/lon from neighbors
        zip_int = int(zip_code)
        neighbors = df_geo[df_geo["geolocation_zip_code_prefix"].astype(int).between(zip_int - window, zip_int + window)]
        if not neighbors.empty:
            lat = neighbors["lat"].median()
            lng = neighbors["lng"].median()
        else:
            lat, lng = None, None

        summary_rows.append({
            "geolocation_zip_code_prefix": zip_code,
            "lat": lat,
            "lng": lng,
            "city": city,
            "state": state
        })

    return pd.DataFrame(summary_rows)


# -------------------------------
# Step 5: Full Workflow
# -------------------------------
# Clean geolocation
df_geolocations_clean = clean_geolocation(df_geolocation)

# Aggregate by ZIP
df_geolocations_clean_zip = aggregate_geolocation(df_geolocations_clean)

# Align with customers
customer_zips = set(df_customers_clean["customer_zip_code_prefix"].unique())
geo_zips = set(df_geolocations_clean_zip["geolocation_zip_code_prefix"].unique())

missing_zips_customers = customer_zips - geo_zips

print("Missing ZIPs (customers):", sorted(missing_zips_customers))

# Patch missing customer ZIPs
summary_df_customers = build_summary(df_customers_clean, df_geolocations_clean_zip)
df_geolocations_clean_zip_updated = pd.concat([df_geolocations_clean_zip, summary_df_customers], ignore_index=True)

# Align with sellers
df_sellers_clean["seller_zip_code_prefix"] = df_sellers_clean["seller_zip_code_prefix"].astype(str).str.zfill(5)
missing_zips_sellers = set(df_sellers_clean["seller_zip_code_prefix"].unique()) - set(df_geolocations_clean_zip_updated["geolocation_zip_code_prefix"].unique())

print("Missing ZIPs (sellers):", sorted(missing_zips_sellers))

# Patch missing seller ZIPs
summary_df_sellers = build_summary_sellers(df_sellers_clean, df_geolocations_clean_zip_updated, missing_zips_sellers)
df_geolocations_clean_zip_updated = pd.concat([df_geolocations_clean_zip_updated, summary_df_sellers], ignore_index=True)

# Final result
print("Final geolocation shape:", df_geolocations_clean_zip_updated.shape)
print(df_geolocations_clean_zip_updated.head(10))
```

```python

def clean_order_items(df_order_items: pd.DataFrame) -> pd.DataFrame:
    """
    Cleans the Order Items dataset and returns a single DataFrame ready for SQL loading:
    - Converts shipping_limit_date to datetime
    - Ensures price and freight_value are numeric
    - Removes invalid prices and freight values
    - Caps extreme values at 99th percentile
    - Preserves all original columns (including order_item_id, order_id, etc.)
    """

    df_clean = df_order_items.copy()

    # 1. Convert shipping_limit_date to datetime
    df_clean["shipping_limit_date"] = pd.to_datetime(
        df_clean["shipping_limit_date"], errors="coerce"
    )

    # 2. Ensure price and freight_value are numeric
    df_clean["price"] = pd.to_numeric(df_clean["price"], errors="coerce")
    df_clean["freight_value"] = pd.to_numeric(df_clean["freight_value"], errors="coerce")

    # 3. Remove invalid prices (<=0)
    df_clean = df_clean[df_clean["price"] > 0]

    # 4. Cap extreme prices at 99th percentile
    cap_price = df_clean["price"].quantile(0.99)
    df_clean["price"] = df_clean["price"].clip(upper=cap_price)

    # 5. Remove invalid freight values (negative)
    df_clean = df_clean[df_clean["freight_value"] >= 0]

    # 6. Cap extreme freight values at 99th percentile
    cap_freight = df_clean["freight_value"].quantile(0.99)
    df_clean["freight_value"] = df_clean["freight_value"].clip(upper=cap_freight)

    # ✅ Return single cleaned DataFrame ready for SQL
    return df_clean



df_order_items_clean = clean_order_items(df_order_items)

print(df_order_items_clean[["price","freight_value"]].describe())
print(df_order_items_clean.shape)
print(df_order_items_clean.dtypes)
```

```python
import numpy as np

def clean_products(df_products: pd.DataFrame, df_order_items: pd.DataFrame) -> pd.DataFrame:
    """
    Cleans the Products dataset and returns a single DataFrame ready for SQL loading:
    - Handles missing category names and product metadata
    - Imputes dimensions with category medians
    - Converts numeric columns to Int64 (nullable integer)
    - Clips unrealistic ranges for length
    - Adjusts zero values for weight and photos
    - Preserves schema with required columns only
    """

    df_clean = df_products.copy()

    # 1. Check if missing products appear in order_items
    missing_products = df_clean[df_clean["product_category_name"].isnull()]["product_id"]
    used_in_orders = df_order_items["product_id"].isin(missing_products).sum()
    print("Missing products used in orders:", used_in_orders)

    # 2. Fill missing category with 'unknown'
    df_clean["product_category_name"] = df_clean["product_category_name"].fillna("unknown")

    # 3. Fill missing name/description/photos with 0
    df_clean["product_name_lenght"] = df_clean["product_name_lenght"].fillna(0)
    df_clean["product_description_lenght"] = df_clean["product_description_lenght"].fillna(0)
    df_clean["product_photos_qty"] = df_clean["product_photos_qty"].fillna(0)

    # 4. Fill missing dimensions with category median
    dims = ["product_weight_g","product_length_cm","product_height_cm","product_width_cm"]
    df_clean[dims] = (
        df_clean.groupby("product_category_name")[dims]
        .transform(lambda x: x.fillna(x.median()))
    )

    # 5. Convert float columns to int
    cols_to_int = [
        "product_name_lenght",
        "product_description_lenght",
        "product_photos_qty",
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ]
    df_clean[cols_to_int] = df_clean[cols_to_int].astype("int64")

    # 6. Replace zero weights with median
    df_clean["product_weight_g"] = df_clean["product_weight_g"].replace(0, np.nan)
    df_clean["product_weight_g"] = df_clean["product_weight_g"].fillna(df_clean["product_weight_g"].median())

    # 7. Clip product_length_cm to realistic range
    df_clean["product_length_cm"] = df_clean["product_length_cm"].clip(lower=5, upper=200)

    # 8. Replace zero photos with 1
    df_clean["product_photos_qty"] = df_clean["product_photos_qty"].replace(0, 1)

    # 9. Keep only required columns
    df_final = df_clean[[
        "product_id",
        "product_category_name",
        "product_weight_g",
        "product_length_cm",
        "product_photos_qty"
    ]].copy()

    # 10. Convert numeric columns safely to Int64 (nullable integer)
    for col in ["product_weight_g", "product_length_cm", "product_photos_qty"]:
        df_final[col] = pd.to_numeric(df_final[col], errors="coerce").astype("Int64")

    # 11. Ensure string columns
    df_final["product_id"] = df_final["product_id"].astype(str)
    df_final["product_category_name"] = df_final["product_category_name"].astype(str)

    # ✅ Return cleaned DataFrame ready for SQL
    return df_final


df_products_clean_sql = clean_products(df_products, df_order_items)

print(df_products_clean_sql.dtypes)
print(df_products_clean_sql.shape)
```

```python
def clean_orders(df_orders: pd.DataFrame) -> pd.DataFrame:
    """
    Cleans the Orders dataset and returns a single DataFrame ready for SQL loading:
    - Converts timestamp columns to datetime
    - Imputes missing approval, carrier, and customer delivery dates using median lags
    - Caps extreme delivery delays and estimated delivery ranges
    - Fixes impossible sequences (delivery before purchase, estimate before purchase)
    - Retains only required columns with proper datatypes
    """

    df_clean = df_orders.copy()

    # 1. Convert string columns to datetime
    datetime_cols = [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]
    for col in datetime_cols:
        df_clean[col] = pd.to_datetime(df_clean[col], errors="coerce")

    # 2. Impute missing approvals (delivered orders only)
    median_lag = (df_clean["order_approved_at"] - df_clean["order_purchase_timestamp"]).median()
    mask = (df_clean["order_approved_at"].isna()) & (df_clean["order_status"] == "delivered")
    df_clean.loc[mask, "order_approved_at"] = df_clean.loc[mask, "order_purchase_timestamp"] + median_lag

    # 3. Impute missing carrier dates (delivered orders only)
    median_shipping_lag = (
        df_clean.loc[df_clean["order_status"] == "delivered", "order_delivered_carrier_date"]
        - df_clean.loc[df_clean["order_status"] == "delivered", "order_purchase_timestamp"]
    ).median()
    mask = (df_clean["order_delivered_carrier_date"].isna()) & (df_clean["order_status"] == "delivered")
    df_clean.loc[mask, "order_delivered_carrier_date"] = (
        df_clean.loc[mask, "order_purchase_timestamp"] + median_shipping_lag
    )

    # 4. Impute missing customer delivery dates (delivered orders only)
    median_delivery_lag = (
        df_clean.loc[df_clean["order_status"] == "delivered", "order_delivered_customer_date"]
        - df_clean.loc[df_clean["order_status"] == "delivered", "order_delivered_carrier_date"]
    ).median()
    mask = (df_clean["order_delivered_customer_date"].isna()) & (df_clean["order_status"] == "delivered")
    df_clean.loc[mask, "order_delivered_customer_date"] = (
        df_clean.loc[mask, "order_delivered_carrier_date"] + median_delivery_lag
    )

    # 5. Delivery days calculation
    df_clean["delivery_days"] = (
        df_clean["order_delivered_customer_date"] - df_clean["order_purchase_timestamp"]
    ).dt.days

    # 6. Cap extreme delays at 60 days
    df_clean.loc[df_clean["delivery_days"] > 60, "order_delivered_customer_date"] = (
        df_clean["order_purchase_timestamp"] + pd.Timedelta(days=60)
    )

    # 7. Fix impossible delivery (before purchase)
    df_clean.loc[
        df_clean["order_delivered_customer_date"] < df_clean["order_purchase_timestamp"],
        "order_delivered_customer_date"
    ] = df_clean["order_purchase_timestamp"]

    # 8. Cap estimated delivery (max 30 days after purchase)
    df_clean.loc[
        df_clean["order_estimated_delivery_date"] > df_clean["order_purchase_timestamp"] + pd.Timedelta(days=30),
        "order_estimated_delivery_date"
    ] = df_clean["order_purchase_timestamp"] + pd.Timedelta(days=30)

    # 9. Fix impossible estimate (before purchase → set to +7 days)
    df_clean.loc[
        df_clean["order_estimated_delivery_date"] < df_clean["order_purchase_timestamp"],
        "order_estimated_delivery_date"
    ] = df_clean["order_purchase_timestamp"] + pd.Timedelta(days=7)

    # 10. Keep only required columns
    df_final = df_clean[[
        "order_id",
        "customer_id",
        "order_status",
        "order_purchase_timestamp",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ]].copy()

    # 11. Enforce datatypes
    df_final["order_id"] = df_final["order_id"].astype(str)
    df_final["customer_id"] = df_final["customer_id"].astype(str)
    df_final["order_status"] = df_final["order_status"].astype(str)

    for col in ["order_purchase_timestamp", "order_delivered_customer_date", "order_estimated_delivery_date"]:
        df_final[col] = pd.to_datetime(df_final[col], errors="coerce")

    return df_final

df_orders_clean_final_sql = clean_orders(df_orders)

print(df_orders_clean_final_sql.dtypes)
print(df_orders_clean_final_sql.shape)
```
